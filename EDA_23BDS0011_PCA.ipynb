```python

#Chirag Rishit
#23BDS0011
# PRINCIPAL COMPONENT ANALYSIS (PCA) - CRIME DATASET


# 1. IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

import warnings
warnings.filterwarnings("ignore")

sns.set_style("whitegrid")



# 2. UPLOAD DATASET FROM PC


uploaded = files.upload()

file_name = next(iter(uploaded))

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())


# 3. BASIC DATASET INFORMATION


print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())



# 4. SELECT NUMERICAL VARIABLES


preferred_columns = [
    "crmrte",
    "prbarr",
    "prbconv",
    "prbpris",
    "avgsen",
    "polpc",
    "density",
    "taxpc",
    "pctmin",
    "pctymle"
]

selected_columns = [
    col for col in preferred_columns
    if col in df.columns
]

# Fallback if preferred columns are not found
if len(selected_columns) < 2:

    numeric_columns = df.select_dtypes(
        include=np.number
    ).columns.tolist()

    excluded_columns = [
        "rownames", "rowname", "id", "index"
    ]

    selected_columns = [
        col for col in numeric_columns
        if col.lower() not in excluded_columns
    ]

if len(selected_columns) < 2:
    raise ValueError("At least two numerical variables are required.")

print("\nSelected Variables:")
print(selected_columns)

pca_df = df[selected_columns].copy()



# 5. HANDLE MISSING AND INFINITE VALUES


pca_df = pca_df.replace([np.inf, -np.inf], np.nan)

# Replace missing values with median
pca_df = pca_df.fillna(pca_df.median())

# Remove constant columns
pca_df = pca_df.loc[:, pca_df.nunique() > 1]

if pca_df.shape[1] < 2:
    raise ValueError("At least two non-constant variables are required.")

print("\nData after preprocessing:", pca_df.shape)

print("\nMissing Values After Cleaning:")
print(pca_df.isnull().sum())



# 6. STANDARDIZATION


scaler = StandardScaler()

scaled_data = scaler.fit_transform(pca_df)

scaled_df = pd.DataFrame(
    scaled_data,
    columns=pca_df.columns
)

print("\nStandardized Dataset:")
display(scaled_df.head())



# 7. APPLY PCA


pca = PCA()

pca_result = pca.fit_transform(scaled_data)

pca_columns = [
    f"PC{i+1}" for i in range(pca_result.shape[1])
]

pca_result_df = pd.DataFrame(
    pca_result,
    columns=pca_columns
)

print("\nPCA Transformed Dataset:")
display(pca_result_df.head())



# 8. EXPLAINED VARIANCE


explained_variance = pca.explained_variance_ratio_

cumulative_variance = np.cumsum(explained_variance)

variance_table = pd.DataFrame({
    "Principal Component": pca_columns,
    "Explained Variance": explained_variance,
    "Explained Variance (%)": explained_variance * 100,
    "Cumulative Variance (%)": cumulative_variance * 100
})

print("\nExplained Variance:")
display(variance_table.round(4))



# GRAPH 1: SCREE PLOT


plt.figure(figsize=(9, 5))

plt.plot(
    range(1, len(explained_variance) + 1),
    explained_variance * 100,
    marker="o"
)

plt.title("Scree Plot - PCA")
plt.xlabel("Principal Components")
plt.ylabel("Explained Variance (%)")
plt.xticks(range(1, len(explained_variance) + 1))
plt.grid(True)
plt.show()



# GRAPH 2: CUMULATIVE EXPLAINED VARIANCE


plt.figure(figsize=(9, 5))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance * 100,
    marker="o"
)

plt.axhline(
    y=95,
    linestyle="--",
    label="95% Variance"
)

plt.title("Cumulative Explained Variance")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Variance (%)")
plt.xticks(range(1, len(cumulative_variance) + 1))
plt.legend()
plt.grid(True)
plt.show()



# GRAPH 3: PCA SCATTER PLOT (PC1 VS PC2)


plt.figure(figsize=(9, 6))

plt.scatter(
    pca_result_df["PC1"],
    pca_result_df["PC2"],
    alpha=0.7
)

plt.title("PCA Scatter Plot (PC1 vs PC2)")
plt.xlabel(
    f"PC1 ({explained_variance[0]*100:.2f}% variance)"
)
plt.ylabel(
    f"PC2 ({explained_variance[1]*100:.2f}% variance)"
)

plt.grid(True)
plt.show()



# 9. PCA LOADINGS


loadings = pd.DataFrame(
    pca.components_.T,
    columns=pca_columns,
    index=pca_df.columns
)

print("\nPCA Loadings:")
display(loadings.round(4))



# GRAPH 4: PCA LOADINGS HEATMAP


plt.figure(figsize=(12, 7))

sns.heatmap(
    loadings,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)

plt.title("PCA Loadings Heatmap")
plt.xlabel("Principal Components")
plt.ylabel("Original Variables")
plt.tight_layout()
plt.show()



# GRAPH 5: INDIVIDUAL EXPLAINED VARIANCE


plt.figure(figsize=(9, 5))

plt.bar(
    pca_columns,
    explained_variance * 100
)

plt.title("Variance Explained by Each Component")
plt.xlabel("Principal Components")
plt.ylabel("Explained Variance (%)")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.show()



# 10. SELECT COMPONENTS EXPLAINING AT LEAST 95% VARIANCE


n_components_95 = np.argmax(cumulative_variance >= 0.95) + 1

print("\nNumber of components required for 95% variance:")
print(n_components_95)

print(
    "Cumulative variance explained:",
    round(cumulative_variance[n_components_95 - 1] * 100, 2),
    "%"
)



# 11. FINAL REDUCED PCA DATASET


pca_final = PCA(n_components=n_components_95)

final_result = pca_final.fit_transform(scaled_data)

final_pca_df = pd.DataFrame(
    final_result,
    columns=[
        f"PC{i+1}" for i in range(n_components_95)
    ]
)

print("\nFinal Reduced PCA Dataset:")
display(final_pca_df.head())

print("\nFinal PCA Dataset Shape:")
print(final_pca_df.shape)




print("\nPCA ANALYSIS COMPLETED SUCCESSFULLY!")

```
