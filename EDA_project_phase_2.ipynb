```python
# Chirag Rishit
#23BDS0011
# PHASE 2
#Uploding a seprate file for this as storage error in the file
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import dendrogram, linkage

sns.set_style("whitegrid")



print("=" * 60)
print("UPLOAD CRIME DATASET")
print("=" * 60)

uploaded = files.upload()

# Get uploaded filename
filename = list(uploaded.keys())[0]

print("\nUploaded file:", filename)

# Read CSV into dataframe
df = pd.read_csv(filename)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)



# 2. BASIC DATASET INFORMATION


print("=" * 60)
print("BASIC DATASET INFORMATION")
print("=" * 60)

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nNumber of rows and columns:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
display(df.isnull().sum())



# 1-D STATISTICAL ANALYSIS


print("=" * 60)
print("1-D STATISTICAL ANALYSIS")
print("=" * 60)


# Important numerical variables from the Crime dataset
variables = [
    "crmrte",
    "prbarr",
    "prbconv",
    "prbpris",
    "avgsen",
    "polpc",
    "density"
]



# Descriptive Statistics


print("\nDescriptive Statistics:")

display(
    df[variables].describe()
)



# Mean


print("\nMean:")

display(
    df[variables].mean()
)



# Median


print("\nMedian:")

display(
    df[variables].median()
)



# Variance


print("\nVariance:")

display(
    df[variables].var()
)



# Standard Deviation


print("\nStandard Deviation:")

display(
    df[variables].std()
)



# 1-D Visualization 1: Histogram


plt.figure(figsize=(8, 5))

sns.histplot(
    df["crmrte"],
    bins=25,
    kde=True
)

plt.title("Distribution of Crime Rate")
plt.xlabel("Crime Rate")
plt.ylabel("Frequency")

plt.show()



# 1-D Visualization 2: Boxplot


plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df["crmrte"]
)

plt.title("Boxplot of Crime Rate")
plt.xlabel("Crime Rate")

plt.show()



# 1-D Visualization 3: Police Population Distribution


plt.figure(figsize=(8, 5))

sns.histplot(
    df["polpc"],
    bins=25,
    kde=True
)

plt.title("Distribution of Police per Population")
plt.xlabel("Police per Population")
plt.ylabel("Frequency")

plt.show()



# 2-D STATISTICAL ANALYSIS


print("=" * 60)
print("2-D STATISTICAL ANALYSIS")
print("=" * 60)



# 2-D Visualization 1
# Crime Rate vs Police per Population


plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="polpc",
    y="crmrte"
)

plt.title("Crime Rate vs Police per Population")
plt.xlabel("Police per Population")
plt.ylabel("Crime Rate")

plt.show()



# 2-D Visualization 2
# Crime Rate vs Probability of Arrest


plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="prbarr",
    y="crmrte"
)

plt.title("Crime Rate vs Probability of Arrest")
plt.xlabel("Probability of Arrest")
plt.ylabel("Crime Rate")

plt.show()



# 2-D Visualization 3
# Crime Rate vs Population Density


plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="density",
    y="crmrte"
)

plt.title("Crime Rate vs Population Density")
plt.xlabel("Population Density")
plt.ylabel("Crime Rate")

plt.show()



# Correlation with Crime Rate


print("\nCorrelation with Crime Rate:")

crime_correlation = (
    df[variables]
    .corr()["crmrte"]
    .sort_values(ascending=False)
)

display(crime_correlation)



# Correlation Heatmap


plt.figure(figsize=(10, 7))

sns.heatmap(
    df[variables].corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()



# 3-D STATISTICAL ANALYSIS


print("=" * 60)
print("3-D STATISTICAL ANALYSIS")
print("=" * 60)


# 3-D Visualization 1
# Police + Arrest Probability + Crime Rate


fig = plt.figure(figsize=(9, 7))

ax = fig.add_subplot(
    111,
    projection="3d"
)

ax.scatter(
    df["polpc"],
    df["prbarr"],
    df["crmrte"],
    c=df["crmrte"],
    cmap="viridis"
)

ax.set_xlabel("Police per Population")
ax.set_ylabel("Probability of Arrest")
ax.set_zlabel("Crime Rate")

ax.set_title(
    "3-D Analysis: Police, Arrest Probability and Crime Rate"
)

plt.show()


# 3-D Visualization 2
# Density + Police + Crime Rate


fig = plt.figure(figsize=(9, 7))

ax = fig.add_subplot(
    111,
    projection="3d"
)

ax.scatter(
    df["density"],
    df["polpc"],
    df["crmrte"],
    c=df["density"],
    cmap="plasma"
)

ax.set_xlabel("Population Density")
ax.set_ylabel("Police per Population")
ax.set_zlabel("Crime Rate")

ax.set_title(
    "3-D Analysis: Density, Police and Crime Rate"
)

plt.show()


# 3-D Visualization 3
# Arrest + Conviction + Crime Rate


fig = plt.figure(figsize=(9, 7))

ax = fig.add_subplot(
    111,
    projection="3d"
)

ax.scatter(
    df["prbarr"],
    df["prbconv"],
    df["crmrte"],
    c=df["crmrte"],
    cmap="coolwarm"
)

ax.set_xlabel("Probability of Arrest")
ax.set_ylabel("Probability of Conviction")
ax.set_zlabel("Crime Rate")

ax.set_title(
    "3-D Analysis: Arrest, Conviction and Crime Rate"
)

plt.show()



# K-MEANS CLUSTERING


print("=" * 60)
print("K-MEANS CLUSTERING")
print("=" * 60)


# Variables selected for clustering
cluster_features = [
    "crmrte",
    "prbarr",
    "prbconv",
    "prbpris",
    "avgsen",
    "polpc",
    "density"
]


cluster_data = df[cluster_features].copy()



# Check missing values before clustering


print("\nMissing values in clustering variables:")

display(
    cluster_data.isnull().sum()
)



# Remove rows containing missing values


cluster_data = cluster_data.dropna()

print("\nClustering dataset shape after removing missing values:")

print(cluster_data.shape)



# Standardization


scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    cluster_data
)

print("\nData standardized successfully.")


# Elbow Method


inertia = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    inertia.append(
        kmeans.inertia_
    )


plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    inertia,
    marker="o"
)

plt.title("Elbow Method for K-Means")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")

plt.show()



# Silhouette Score


silhouette_scores = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(
        X_scaled
    )

    score = silhouette_score(
        X_scaled,
        labels
    )

    silhouette_scores.append(
        score
    )


plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    silhouette_scores,
    marker="o"
)

plt.title("Silhouette Score for K-Means")
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")

plt.show()



# Apply K-Means with 3 Clusters


kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans.fit_predict(
    X_scaled
)


# Add results only to the rows used for clustering
cluster_data["KMeans_Cluster"] = kmeans_labels


print("\nK-Means Cluster Counts:")

print(
    cluster_data["KMeans_Cluster"]
    .value_counts()
    .sort_index()
)



# K-Means Visualization


plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=cluster_data,
    x="polpc",
    y="crmrte",
    hue="KMeans_Cluster",
    palette="viridis",
    s=70
)

plt.title("K-Means Clustering")
plt.xlabel("Police per Population")
plt.ylabel("Crime Rate")

plt.show()



# K-Means Cluster Profile


print("\nK-Means Cluster Profile:")

kmeans_profile = (
    cluster_data
    .groupby("KMeans_Cluster")[cluster_features]
    .mean()
)

display(
    kmeans_profile
)



# K-Means Silhouette Score


kmeans_silhouette = silhouette_score(
    X_scaled,
    kmeans_labels
)

print(
    "\nK-Means Silhouette Score:",
    round(kmeans_silhouette, 4)
)



# HIERARCHICAL CLUSTERING


print("=" * 60)
print("HIERARCHICAL CLUSTERING")
print("=" * 60)



# Create linkage matrix


linked = linkage(
    X_scaled,
    method="ward"
)



# Dendrogram


plt.figure(figsize=(14, 7))

dendrogram(
    linked,
    truncate_mode="lastp",
    p=30
)

plt.title(
    "Hierarchical Clustering Dendrogram"
)

plt.xlabel("Clusters / Observations")
plt.ylabel("Distance")

plt.show()



# Agglomerative Clustering


hierarchical = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

hierarchical_labels = hierarchical.fit_predict(
    X_scaled
)


# Add result only to clustering dataframe
cluster_data["Hierarchical_Cluster"] = hierarchical_labels


print("\nHierarchical Cluster Counts:")

print(
    cluster_data["Hierarchical_Cluster"]
    .value_counts()
    .sort_index()
)


# Hierarchical Visualization


plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=cluster_data,
    x="polpc",
    y="crmrte",
    hue="Hierarchical_Cluster",
    palette="Set1",
    s=70
)

plt.title("Hierarchical Clustering")
plt.xlabel("Police per Population")
plt.ylabel("Crime Rate")

plt.show()



# Hierarchical Cluster Profile


print("\nHierarchical Cluster Profile:")

hierarchical_profile = (
    cluster_data
    .groupby("Hierarchical_Cluster")[cluster_features]
    .mean()
)

display(
    hierarchical_profile
)



# Hierarchical Silhouette Score


hierarchical_silhouette = silhouette_score(
    X_scaled,
    hierarchical_labels
)

print(
    "\nHierarchical Silhouette Score:",
    round(hierarchical_silhouette, 4)
)



# FINAL COMPARISON


print("=" * 60)
print("CLUSTERING COMPARISON")
print("=" * 60)


comparison = pd.DataFrame({
    "Method": [
        "K-Means",
        "Hierarchical"
    ],
    "Silhouette Score": [
        kmeans_silhouette,
        hierarchical_silhouette
    ]
})


display(comparison)




print("\nPhase 2 analysis completed.")
```
