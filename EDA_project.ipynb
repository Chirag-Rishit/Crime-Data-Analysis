```python
#NAME:CHIRAG RISHIT
#REG.NO:23BDS0011
#SUBJECT:EDA



import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder

plt.style.use("ggplot")
sns.set(font_scale=1)



from google.colab import files

uploaded = files.upload()

# Change filename if required
df = pd.read_csv("Crime.csv")



print("="*60)
print("DATASET INFORMATION")
print("="*60)

print("\nShape of Dataset:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst Five Records:")
display(df.head())


# BASIC STATISTICAL ANALYSIS


print("="*60)
print("STATISTICAL ANALYSIS")
print("="*60)

display(df.describe(include='all'))

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Records")
print(df.duplicated().sum())


#  HANDLE MISSING VALUES


print("="*60)
print("HANDLING MISSING VALUES")
print("="*60)

numeric_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(include='object').columns

for col in numeric_cols:
    df[col].fillna(df[col].median(), inplace=True)

for col in categorical_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)

print("Missing Values After Cleaning")
print(df.isnull().sum())


#  DATA CLEANING


print("="*60)
print("DATA CLEANING")
print("="*60)

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Remove spaces from text columns
for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip()

print("Dataset Shape After Cleaning:", df.shape)


# DATA TRANSFORMATION


print("="*60)
print("DATA TRANSFORMATION")
print("="*60)

encoder = LabelEncoder()

for col in categorical_cols:
    df[col] = encoder.fit_transform(df[col])

print("Categorical Columns Converted Successfully")

display(df.head())


# UNIVARIATE ANALYSIS


print("="*60)
print("UNIVARIATE ANALYSIS")
print("="*60)

# Histogram

df.hist(figsize=(16,12), bins=20)

plt.suptitle("Histogram of Numerical Features", fontsize=18)

plt.show()

# Boxplot

plt.figure(figsize=(16,8))

sns.boxplot(data=df)

plt.xticks(rotation=90)

plt.title("Boxplot of Dataset")

plt.show()

# Distribution Plot

if len(numeric_cols)>0:

    plt.figure(figsize=(10,6))

    sns.histplot(df[numeric_cols[0]], kde=True, color='blue')

    plt.title(f"Distribution of {numeric_cols[0]}")

    plt.show()


# BIVARIATE ANALYSIS

print("="*60)
print("BIVARIATE ANALYSIS")
print("="*60)

if len(numeric_cols)>=2:

    # Scatter Plot

    plt.figure(figsize=(8,6))

    sns.scatterplot(
        x=df[numeric_cols[0]],
        y=df[numeric_cols[1]],
        color='red'
    )

    plt.title(f"{numeric_cols[0]} vs {numeric_cols[1]}")

    plt.show()

    # Regression Plot

    plt.figure(figsize=(8,6))

    sns.regplot(
        x=df[numeric_cols[0]],
        y=df[numeric_cols[1]],
        scatter_kws={"color":"green"},
        line_kws={"color":"black"}
    )

    plt.title("Regression Plot")

    plt.show()

# Correlation Heatmap

plt.figure(figsize=(12,8))

sns.heatmap(df.corr(),
            annot=True,
            cmap="coolwarm",
            linewidths=0.5)

plt.title("Correlation Heatmap")

plt.show()


# MULTIVARIATE ANALYSIS


print("="*60)
print("MULTIVARIATE ANALYSIS")
print("="*60)

# Pair Plot

if len(numeric_cols)>=4:

    sns.pairplot(df[numeric_cols[:4]])

    plt.show()

# Cluster Heatmap

sns.clustermap(df.corr(),
               cmap="viridis",
               figsize=(10,10))

plt.show()

# Correlation Matrix

plt.figure(figsize=(12,10))

mask = np.triu(np.ones_like(df.corr(), dtype=bool))

sns.heatmap(df.corr(),
            mask=mask,
            cmap="RdYlBu",
            annot=False)

plt.title("Correlation Matrix")

plt.show()


# FINAL DATASET


print("="*60)
print("FINAL CLEANED DATASET")
print("="*60)

display(df.head())

print("Final Shape:", df.shape)

# -----------------------------
# SAVE CLEAN DATA
# -----------------------------

df.to_csv("Crime_Cleaned.csv", index=False)

print("\nCleaned dataset saved as Crime_Cleaned.csv")

files.download("Crime_Cleaned.csv")

print("\nPROJECT COMPLETED SUCCESSFULLY")
```
